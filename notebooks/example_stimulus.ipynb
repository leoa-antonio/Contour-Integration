# Example Stimulus Generation

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent / "src"))

import numpy as np
from matplotlib.pyplot import subplots
from stimuli import generate_stimulus, plot_example_stimuli
```

## Generate Example Stimuli

```python
seed = 0
rng = np.random.default_rng(seed)

# Example Values from Field, Hayes, Hess (1993)
beta = np.deg2rad(30)
step_length = 32

positive_stimulus = generate_stimulus(rng, beta, step_length, contour_present=True)
negative_stimulus = generate_stimulus(rng, beta, step_length, contour_present=False)
```

## Plot

```python
plot_example_stimuli(positive_stimulus, negative_stimulus)
```
